```python
def cleaning(sentence):
  sentence=sentence.lower()

  #Remove leading/trailing spaces
  sentence=sentence.strip()

  #Normalize multiple spaces
  sentence=" ".join(sentence.split())

  return sentence
```

```python
#custom dataset class

class CustomDataset(Dataset):
  def __init__(self,en_id,fr_id):
    super().__init__()
    self.en_id=en_id
    self.fr_id=fr_id

  def __len__(self):
    return len(self.en_id)

  def __getitem__(self,idx):
    en=self.en_id[idx]
    fr=self.fr_id[idx]

    return en,fr
```

```python
# collate func -> padding to make same length

def collate_fn(batch):

  max_en_len=0
  max_fr_len=0

  for en,fr in batch:
    max_en_len=max(max_en_len,len(en))
    max_fr_len=max(max_fr_len,len(fr))

  en_pad=[]
  fr_pad=[]

  for en,fr in batch:
    en_padded=en+[en_word2idx["<PAD>"]]*(max_en_len-len(en))
    fr_padded=fr+[fr_word2idx["<PAD>"]]*(max_fr_len-len(fr))

    en_pad.append(en_padded)
    fr_pad.append(fr_padded)

  en_pad=torch.tensor(en_pad,dtype=torch.long)
  fr_pad=torch.tensor(fr_pad,dtype=torch.long)

  return en_pad,fr_pad
```

```python
#create train and test loader
train_loader = DataLoader(train_dataset,batch_size=batch_size,shuffle=True,collate_fn=collate_fn,pin_memory=True,num_workers=0)

val_loader = DataLoader(val_dataset,batch_size=batch_size,shuffle=False,collate_fn=collate_fn,pin_memory=True)
```

## Architecture

```python
class ForgetGate(nn.Module):
  def __init__(self,input_size, hidden_size):
    super().__init__()
    self.forget_gate = nn.Linear(input_size + hidden_size, hidden_size, dtype=torch.float32)

  def forward(self,x_t,h_t_1):
    concat=torch.cat([x_t,h_t_1],dim=1)
    f_t=torch.sigmoid(self.forget_gate(concat))
    return f_t
```

```python
class InputGate(nn.Module):
  def __init__(self,input_size, hidden_size):
    super().__init__()
    self.input_gate1 = nn.Linear(input_size + hidden_size, hidden_size, dtype=torch.float32)
    self.input_gate2 = nn.Linear(input_size + hidden_size, hidden_size, dtype=torch.float32)

  def forward(self,x_t,h_t_1):
    concat=torch.cat([x_t,h_t_1],dim=1)
    i_t=torch.sigmoid(self.input_gate1(concat))
    ct_bar=torch.tanh(self.input_gate2(concat))
    return i_t, ct_bar
```

```python
class OutputGate(nn.Module):
  def __init__(self,input_size, hidden_size):
    super().__init__()
    self.output_gate1 = nn.Linear(input_size + hidden_size, hidden_size, dtype=torch.float32)

  def forward(self,x_t,h_t_1):
    concat=torch.cat([x_t,h_t_1],dim=1)
    o_t=torch.sigmoid(self.output_gate1(concat))
    return o_t
```

```python
class LSTMBlock(nn.Module):
  def __init__(self,input_size, hidden_size):
    super().__init__()
    self.input_gate = InputGate(input_size=input_size, hidden_size=hidden_size)
    self.output_gate = OutputGate(input_size=input_size, hidden_size=hidden_size)
    self.forget_gate = ForgetGate(input_size=input_size, hidden_size=hidden_size)

  def forward(self,x_t,h_t_1,c_t_1):
    x_t = x_t.squeeze(1) if x_t.dim() == 3 else x_t

    f_t=self.forget_gate(x_t,h_t_1)
    i_t, ct_bar=self.input_gate(x_t,h_t_1)
    o_t=self.output_gate(x_t,h_t_1)

    c_t=c_t_1*f_t + i_t*ct_bar
    h_t=o_t*torch.tanh(c_t)

    return h_t,c_t
```

```python
class EmbeddingLayer(nn.Module):
  def __init__(self,vocab_size,embedding_dim):
    super().__init__()
    self.embedding=nn.Embedding(vocab_size,embedding_dim)

  def forward(self,x):
    return self.embedding(x)
```

```python
class Encoder(nn.Module):
  def __init__(self,vocab_size,embedding_dim,hidden_size):
    super().__init__()
    self.vocab_size=vocab_size
    self.embedding_dim=embedding_dim
    self.hidden_size=hidden_size

    self.embedding=EmbeddingLayer(vocab_size,embedding_dim)
    self.dropout = nn.Dropout(0.5)
    self.lstm=LSTMBlock(embedding_dim,hidden_size)


  def forward(self,x): # x.shape= batch_size,max_en_len
    x_embed=self.embedding(x) # x_embed.shape= batch_size,max_en_len,embedding_dim
    x_embed = self.dropout(x_embed)
    batch_size=x.shape[0]
    h_t=torch.zeros(batch_size,self.hidden_size,device=x.device)
    c_t=torch.zeros(batch_size,self.hidden_size,device=x.device)

    # print(x.shape)
    # print(x_embed.shape)

    for t in range(x_embed.shape[1]):
      h_t,c_t=self.lstm(x_embed[:,t,:],h_t,c_t)

    return h_t,c_t
```

```python
class Decoder(nn.Module):
  def __init__(self,vocab_size,embedding_dim,hidden_size):
    super().__init__()
    self.vocab_size=vocab_size
    self.embedding_dim=embedding_dim
    self.hidden_size=hidden_size

    self.embedding=EmbeddingLayer(vocab_size,embedding_dim)
    self.dropout = nn.Dropout(0.5)
    self.lstm=LSTMBlock(embedding_dim,hidden_size)
    self.linear=nn.Linear(hidden_size,vocab_size)

  def forward(self,x,h_t_1,c_t_1):
    x_embed=self.embedding(x)
    x_embed = self.dropout(x_embed)
    h_t,c_t=self.lstm(x_embed,h_t_1,c_t_1)
    output=self.linear(h_t)
    return output,h_t,c_t
```

```python
class Seq2Seq(nn.Module):
  def __init__(self,en_vocab_size,fr_vocab_size,embedding_dim,hidden_size):
    super().__init__()
    self.en_vocab_size=en_vocab_size
    self.fr_vocab_size=fr_vocab_size
    self.embedding_dim=embedding_dim
    self.hidden_size=hidden_size

    self.encoder=Encoder(en_vocab_size,embedding_dim,hidden_size)
    self.decoder=Decoder(fr_vocab_size,embedding_dim,hidden_size)

  def forward(self, en_x, fr_x):
    # encoder
    h_t, c_t = self.encoder(en_x)

    outputs = []

    # decoder
    for t in range(fr_x.shape[1]):
        # Get the current token (starts with <SOS> at t=0)
        input = fr_x[:, t].unsqueeze(1)

        output, h_t, c_t = self.decoder(input, h_t, c_t)
        outputs.append(output.unsqueeze(1))

    outputs = torch.cat(outputs, dim=1)
    return outputs
```

```python
# create model
model=Seq2Seq(len(en_word2idx),len(fr_word2idx),embedding_dim,hidden_size)
model=model.to(device)

#loss func
criterion = nn.CrossEntropyLoss(ignore_index=fr_word2idx["<PAD>"])

#define optimizer
optimizer=torch.optim.Adam(model.parameters(),lr=learning_rate)
```

```python
model.train()

train_loss_history = []
final_epoch_loss = []

for epoch in range(epochs):

    epoch_loss = 0

    for batch_idx, (en_batch, fr_batch) in enumerate(train_loader):

        en_batch = en_batch.to(device)
        fr_batch = fr_batch.to(device)

        decoder_input = fr_batch[:, :-1]
        target = fr_batch[:, 1:]

        predictions = model(en_batch, decoder_input)

        B, T_minus_1, V = predictions.shape
        predictions = predictions.view(-1, V)
        target = target.reshape(-1)

        loss = criterion(predictions, target)

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5)
        optimizer.step()

        loss_value = loss.item()
        train_loss_history.append(loss_value)
        epoch_loss += loss_value

        del predictions, loss, target, decoder_input

    # End of epoch logging
    avg_epoch_loss = epoch_loss / len(train_loader)
    final_epoch_loss.append(avg_epoch_loss)

    # NEW: Print only once per epoch
    print(f"Epoch [{epoch+1}/{epochs}] Completed - Average Loss: {avg_epoch_loss:.4f}")
```

```python
import torch

checkpoint = {
    "epoch": epoch,  # current epoch number
    "model_state_dict": model.state_dict(),
    "optimizer_state_dict": optimizer.state_dict(),
    "train_loss_history": train_loss_history,
    "final_epoch_loss": final_epoch_loss,
}

torch.save(checkpoint, "seq2seq_stage0_checkpoint.pth")

print("Checkpoint saved successfully!")
```

```python
def greedy_translate(model, sentence, max_len=60):
    model.eval()

    with torch.no_grad():

        # ----------------------------
        # Convert sentence -> indices
        # ----------------------------
        tokens = sentence.lower().strip().split()

        en_indices = []

        for token in tokens:
            if token in en_word2idx:
                en_indices.append(en_word2idx[token])
            else:
                en_indices.append(en_word2idx["<UNK>"])

        en_tensor = torch.tensor([en_indices], dtype=torch.long).to(device)

        # ----------------------------
        # Encoder
        # ----------------------------
        h_t, c_t = model.encoder(en_tensor)

        # ----------------------------
        # First decoder input = <SOS>
        # ----------------------------
        decoder_input = torch.tensor(
            [fr_word2idx["<SOS>"]],
            dtype=torch.long,
            device=device
        )

        predicted_sentence = []

        # ----------------------------
        # Greedy decoding
        # ----------------------------
        for _ in range(max_len):

            output, h_t, c_t = model.decoder(
                decoder_input,
                h_t,
                c_t
            )

            predicted_idx = output.argmax(dim=1).item()

            if predicted_idx == fr_word2idx["<EOS>"]:
                break

            predicted_sentence.append(
                fr_idx2word.get(predicted_idx, "<UNK>")
            )

            decoder_input = torch.tensor(
                [predicted_idx],
                dtype=torch.long,
                device=device
            )

    return " ".join(predicted_sentence)
```

```python
sentence = "how are you"

translation = greedy_translate(model, sentence)

print("English :", sentence)
print("French  :", translation)
```

```python
def test_translation(model, val_en, val_fr, num_examples=10):

    for i in range(num_examples):

        # -------------------------
        # English sentence
        # -------------------------
        en_sentence = " ".join(
            [en_idx2word[idx] for idx in val_en[i]]
        )

        # -------------------------
        # Ground truth French
        # -------------------------
        fr_sentence = " ".join(
            [
                fr_idx2word[idx]
                for idx in val_fr[i]
                if idx not in (
                    fr_word2idx["<PAD>"],
                    fr_word2idx["<SOS>"],
                    fr_word2idx["<EOS>"]
                )
            ]
        )

        # -------------------------
        # Model prediction
        # -------------------------
        prediction = greedy_translate(model, en_sentence)

        print("=" * 80)
        print("English : ", en_sentence)
        print("Actual   : ", fr_sentence)
        print("Predicted: ", prediction)
        print()
```

```python
test_translation(
    model,
    val_en,
    val_fr,
    num_examples=10
)
```

```python
from sacrebleu import corpus_bleu

def evaluate_bleu(model, val_en, val_fr, num_samples=1000):

    predictions = []
    references = []

    model.eval()

    with torch.no_grad():

        for i in range(min(num_samples, len(val_en))):

            # -------------------------
            # English sentence
            # -------------------------
            en_sentence = " ".join(
                [en_idx2word[idx] for idx in val_en[i]]
            )

            # -------------------------
            # Model prediction
            # -------------------------
            pred = greedy_translate(model, en_sentence)

            predictions.append(pred)

            # -------------------------
            # Reference sentence
            # -------------------------
            ref = " ".join(
                [
                    fr_idx2word[idx]
                    for idx in val_fr[i]
                    if idx not in (
                        fr_word2idx["<PAD>"],
                        fr_word2idx["<SOS>"],
                        fr_word2idx["<EOS>"]
                    )
                ]
            )

            references.append(ref)

    bleu = corpus_bleu(
        predictions,
        [references]
    )

    print(f"BLEU Score : {bleu.score:.2f}")

    return bleu.score
```
