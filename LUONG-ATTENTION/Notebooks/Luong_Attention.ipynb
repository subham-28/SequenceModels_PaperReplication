```python
import re

def cleaning(sentence):
    sentence=re.sub(r"([?.!,;:\'\"()])", r" \1 ", sentence)

    # Normalize multiple spaces
    sentence=" ".join(sentence.split())

    return sentence
```

```python
def preprocess(example):
    example["translation"]["en"]=cleaning(example["translation"]["en"])
    example["translation"]["fr"]=cleaning(example["translation"]["fr"])
    return example

train_dataset=train_dataset.map(preprocess)
```

```python
# for english
en_word2idx={
    "<PAD>":0,
    "<SOS>":1,
    "<EOS>":2,
    "<UNK>":3
}

for i,(word,freq) in enumerate(en_vocab):
  en_word2idx[word]=i+4


# for french
fr_word2idx={
    "<PAD>":0,
    "<SOS>":1,
    "<EOS>":2,
    "<UNK>":3
}

for i,(word,freq) in enumerate(fr_vocab):
  fr_word2idx[word]=i+4
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset,DataLoader
import random
```

```python
#custom dataset class

class CustomDataset(Dataset):
  def __init__(self,en_id,fr_id):
    super().__init__()
    self.en_id=en_id
    self.fr_id=fr_id

  def __len__(self):
    return len(self.en_id)

  def __getitem__(self,idx):
    en=self.en_id[idx]
    fr=self.fr_id[idx]

    return en,fr
```

```python
# collate func -> padding to make same length

def collate_fn(batch):

  max_en_len=0
  max_fr_len=0

  for en,fr in batch:
    max_en_len=max(max_en_len,len(en))
    max_fr_len=max(max_fr_len,len(fr))

  en_pad=[]
  fr_pad=[]

  for en,fr in batch:
    en_padded=en+[en_word2idx["<PAD>"]]*(max_en_len-len(en))
    fr_padded=fr+[fr_word2idx["<PAD>"]]*(max_fr_len-len(fr))

    en_pad.append(en_padded)
    fr_pad.append(fr_padded)

  en_pad=torch.tensor(en_pad,dtype=torch.long)
  fr_pad=torch.tensor(fr_pad,dtype=torch.long)

  return en_pad,fr_pad
```

```python
train_dataset=CustomDataset(train_en,train_fr)
val_dataset=CustomDataset(val_en,val_fr)
```

```python
#create train and test loader
train_loader = DataLoader(train_dataset,batch_size=batch_size,shuffle=True,collate_fn=collate_fn,pin_memory=True,num_workers=0)

val_loader = DataLoader(val_dataset,batch_size=batch_size,shuffle=False,collate_fn=collate_fn,pin_memory=True)
```

### Architecture

```python
class ForgetGate(nn.Module):
  def __init__(self,input_size, hidden_size):
    super().__init__()
    self.forget_gate = nn.Linear(input_size + hidden_size, hidden_size, dtype=torch.float32)

  def forward(self,x_t,h_t_1):
    concat=torch.cat([x_t,h_t_1],dim=1)
    f_t=torch.sigmoid(self.forget_gate(concat))
    return f_t
```

```python
class InputGate(nn.Module):
  def __init__(self,input_size, hidden_size):
    super().__init__()
    self.input_gate1 = nn.Linear(input_size + hidden_size, hidden_size, dtype=torch.float32)
    self.input_gate2 = nn.Linear(input_size + hidden_size, hidden_size, dtype=torch.float32)

  def forward(self,x_t,h_t_1):
    concat=torch.cat([x_t,h_t_1],dim=1)
    i_t=torch.sigmoid(self.input_gate1(concat))
    ct_bar=torch.tanh(self.input_gate2(concat))
    return i_t, ct_bar
```

```python
class OutputGate(nn.Module):
  def __init__(self,input_size, hidden_size):
    super().__init__()
    self.output_gate1 = nn.Linear(input_size + hidden_size, hidden_size, dtype=torch.float32)

  def forward(self,x_t,h_t_1):
    concat=torch.cat([x_t,h_t_1],dim=1)
    o_t=torch.sigmoid(self.output_gate1(concat))
    return o_t
```

```python
class LSTMBlock(nn.Module):
  def __init__(self,input_size, hidden_size):
    super().__init__()
    self.input_gate = InputGate(input_size=input_size, hidden_size=hidden_size)
    self.output_gate = OutputGate(input_size=input_size, hidden_size=hidden_size)
    self.forget_gate = ForgetGate(input_size=input_size, hidden_size=hidden_size)

  def forward(self,x_t,h_t_1,c_t_1):
    x_t = x_t.squeeze(1) if x_t.dim() == 3 else x_t

    f_t=self.forget_gate(x_t,h_t_1)
    i_t, ct_bar=self.input_gate(x_t,h_t_1)
    o_t=self.output_gate(x_t,h_t_1)

    c_t=c_t_1*f_t + i_t*ct_bar
    h_t=o_t*torch.tanh(c_t)

    return h_t,c_t
```

```python
class EmbeddingLayer(nn.Module):
  def __init__(self,vocab_size,embedding_dim):
    super().__init__()
    self.embedding=nn.Embedding(vocab_size,embedding_dim)

  def forward(self,x):
    return self.embedding(x)
```

```python
class LuongEncoder(nn.Module):
  def __init__(self,vocab_size,embedding_dim,hidden_size,num_layers=2,dropout=0.5):
    super().__init__()
    self.hidden_size=hidden_size
    self.num_layers=num_layers
    self.embedding=EmbeddingLayer(vocab_size,embedding_dim)
    self.dropout = nn.Dropout(dropout)
    self.lstm_layers = nn.ModuleList([
            LSTMBlock(embedding_dim if i == 0 else hidden_size, hidden_size)
            for i in range(num_layers)
        ])


  def forward(self,x): # x.shape= batch_size,max_en_len
    x_embed=self.embedding(x) # x_embed.shape= batch_size,max_en_len,embedding_dim
    x_embed = self.dropout(x_embed)
    batch_size, seq_len=x.size()

    h_t=[torch.zeros(batch_size,self.hidden_size,device=x.device) for _ in range(self.num_layers)]
    c_t=[torch.zeros(batch_size,self.hidden_size,device=x.device) for _ in range(self.num_layers)]

    top_layer_states=[]

    # print(x.shape)
    # print(x_embed.shape)

    for t in range(seq_len): # forward RNN
      layer_input = x_embed[:, t, :]
      for l in range(self.num_layers):
        h_t[l],c_t[l]=self.lstm_layers[l](layer_input,h_t[l],c_t[l])
        layer_input = h_t[l]

        if l<self.num_layers-1:
          layer_input = self.dropout(layer_input)

      top_layer_states.append(h_t[-1])

    annotations=torch.stack(top_layer_states, dim=1)

    final_hidden=h_t[-1]
    final_cell=c_t[-1]

    return annotations, final_hidden, final_cell
```

### $$e_{ij} = s_{i}^T .  h_j$$

```python
# The Alignment Model (which computes the attention weights and context vector)

class AlignmentModel(nn.Module):
  def __init__(self,hidden_size):
    super().__init__()
    self.W_a=nn.Linear(hidden_size,hidden_size,bias=False)

  def forward(self,h_t,annotations): # h_t.shape= batch_size,hidden_size   # annotations.shape= batch_size,max_en_len,2*hidden_size
    h_t=h_t.unsqueeze(1) # h_t.shape= batch_size,1,hidden_size
    processed_annotations=self.W_a(annotations) # processed_annotations.shape= batch_size,max_en_len,hidden_size

    e_ij=torch.bmm(h_t,processed_annotations.transpose(1,2)) # e_ij.shape= batch_size,max_en_len,1
    e_ij=e_ij.squeeze(1) # e_ij shape: (batch_size, seq_len, 1) -> squeeze to (batch_size, seq_len)
    alpha_ij=F.softmax(e_ij,dim=1) # alpha_ij shape: (batch_size, seq_len)

    # torch.bmm (Batch Matrix Multiplication) computes the weighted sum efficiently
    c_t=torch.bmm(alpha_ij.unsqueeze(1), annotations).squeeze(1) # Result c_i shape: (batch_size, 1, annotation_size) -> squeeze to (batch_size, annotation_size)
    return c_t, alpha_ij
```

```python
class LuongDecoder(nn.Module):
  def __init__(self,vocab_size,embedding_dim,hidden_size,dropout=0.5):
    super().__init__()
    self.vocab_size=vocab_size
    self.hidden_size=hidden_size
    self.embedding=EmbeddingLayer(vocab_size,embedding_dim)
    self.dropout = nn.Dropout(dropout)

    self.attention=AlignmentModel(hidden_size)
    self.lstm=LSTMBlock(embedding_dim+hidden_size,hidden_size)
    self.W_c=nn.Linear(hidden_size*2,hidden_size,bias=False)

    self.W_s=nn.Linear(hidden_size,vocab_size)

  def forward(self,y_prev,s_prev,c_prev_cell,h_tilde_prev,annotations):

    y_embed=self.embedding(y_prev)
    y_embed=self.dropout(y_embed)

    rnn_input=torch.cat([y_embed,h_tilde_prev],dim=1)

    # 1st step
    h_t,c_t_cell=self.lstm(rnn_input,s_prev,c_prev_cell)

    c_t,alpha_ij=self.attention(h_t,annotations)

    #concat vectors
    concat_vec=torch.cat([c_t,h_t],dim=1)
    h_tilde_t = torch.tanh(self.W_c(concat_vec))

    output = self.W_s(h_tilde_t)
    return output, h_t, c_t_cell, h_tilde_t, alpha_ij
```

```python
class LuongAttention(nn.Module):
    def __init__(self, en_vocab_size, fr_vocab_size, embedding_dim, hidden_size, num_layers=2):
        super().__init__()
        self.en_vocab_size = en_vocab_size
        self.fr_vocab_size = fr_vocab_size
        self.embedding_dim = embedding_dim
        self.hidden_size = hidden_size

        self.encoder = LuongEncoder(vocab_size=en_vocab_size,embedding_dim=embedding_dim,hidden_size=hidden_size,num_layers=num_layers)
        self.decoder = LuongDecoder(vocab_size=fr_vocab_size,embedding_dim=embedding_dim,hidden_size=hidden_size)

    def forward(self, en_x, fr_x, teacher_forcing_ratio=0.5):
        batch_size = en_x.shape[0]
        target_len = fr_x.shape[1]

        # Initialize tensor to store the final predictions from the decoder
        outputs = torch.zeros(batch_size, target_len, self.fr_vocab_size).to(en_x.device)

        # 1. Encode the source sequence[cite: 1]
        annotations, h_t, c_t = self.encoder(en_x)

        # 2. Extract the first input token for the decoder (<SOS> token)
        input_token = fr_x[:, 0]

        h_tilde_t = torch.zeros(batch_size, self.hidden_size, device=en_x.device)

        # 3. Decode step-by-step
        for t in range(1, target_len):

            # Pass the previous token, previous states, and all annotations to the decoder[cite: 1]
            output, h_t, c_t, h_tilde_t, _ = self.decoder(input_token, h_t, c_t, h_tilde_t, annotations)

            # Store the current time step's prediction
            outputs[:, t, :] = output

            # Extract the index of the highest predicted word
            top1 = output.argmax(1)

            # 4. Teacher Forcing
            teacher_force = random.random() < teacher_forcing_ratio
            input_token = fr_x[:, t] if teacher_force else top1

        return outputs
```

```python
# create model
model=LuongAttention(len(en_word2idx),len(fr_word2idx),embedding_dim,hidden_size)
model=model.to(device)

# 2. Loss func
criterion=nn.CrossEntropyLoss(ignore_index=fr_word2idx["<PAD>"])

# 3. Define optimizer (Changed to strictly match Luong et al.)
optimizer=torch.optim.SGD(model.parameters(), lr=1.0)
```

```python
import torch

# Define the scheduler BEFORE the loop
# This will halve the learning rate (gamma=0.5) at the end of epochs 8, 9, 10, and 11[cite: 1].
scheduler = torch.optim.lr_scheduler.MultiStepLR(
    optimizer,
    milestones=[8, 9, 10, 11],
    gamma=0.5
)

model.train()

train_loss_history = []
final_epoch_loss = []

for epoch in range(epochs):

    epoch_loss = 0

    for batch_idx, (en_batch, fr_batch) in enumerate(train_loader):

        en_batch = en_batch.to(device)
        fr_batch = fr_batch.to(device)

        # 1. Pass the entire batch.
        predictions = model(en_batch, fr_batch)

        # 2. Slice off the first time step.
        predictions = predictions[:, 1:, :]
        target = fr_batch[:, 1:]

        # Reshape for CrossEntropyLoss
        B, T_minus_1, V = predictions.shape
        predictions = predictions.reshape(-1, V)
        target = target.reshape(-1)

        loss = criterion(predictions, target)

        optimizer.zero_grad()
        loss.backward()

        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)

        optimizer.step()

        loss_value = loss.item()
        train_loss_history.append(loss_value)
        epoch_loss += loss_value

        del predictions, loss, target

    # End of epoch logging
    avg_epoch_loss = epoch_loss / len(train_loader)
    final_epoch_loss.append(avg_epoch_loss)

    current_lr = optimizer.param_groups[0]['lr']
    print(f"Epoch [{epoch+1}/{epochs}] Completed - Avg Loss: {avg_epoch_loss:.4f} - LR: {current_lr}")

    scheduler.step()
```

```python
import torch

def translate_sentence(model, sentence_ids, en_idx2word, fr_idx2word, device, max_length=50):
    # Set model to evaluation mode (turns off dropout)
    model.eval()

    # Create batch of 1
    source_tensor = torch.tensor([sentence_ids], dtype=torch.long).to(device)

    with torch.no_grad():
        # 1. Encode the source sentence
        annotations, h_t, c_t = model.encoder(source_tensor)

    sos_idx = 1
    eos_idx = 2
    decoder_input = torch.tensor([sos_idx], dtype=torch.long).to(device)

    # FIX 1: Initialize the attentional vector for the first step (batch_size = 1)
    h_tilde_t = torch.zeros(1, model.hidden_size, device=device)

    translated_sentence_ids = []

    # 3. Decode step-by-step
    for _ in range(max_length):
        with torch.no_grad():
            # FIX 2: Pass and receive h_tilde_t to match the LuongDecoder signature
            output, h_t, c_t, h_tilde_t, _ = model.decoder(decoder_input, h_t, c_t, h_tilde_t, annotations)

        # Get the highest predicted word index
        predicted_word_idx = output.argmax(1).item()

        # If the model predicts <EOS>, stop generating
        if predicted_word_idx == eos_idx:
            break

        translated_sentence_ids.append(predicted_word_idx)

        # The predicted word becomes the input for the next time step
        decoder_input = torch.tensor([predicted_word_idx], dtype=torch.long).to(device)

    # Convert integer IDs back to strings
    translated_words = [fr_idx2word[idx] for idx in translated_sentence_ids]

    return translated_words
```

```python
from nltk.translate.bleu_score import corpus_bleu

def calculate_bleu(model, val_en, val_fr, en_idx2word, fr_idx2word, device):
    targets = []
    outputs = []

    for i in range(len(val_en)):
        # 1. Get the source and target integer lists
        source_sentence_ids = val_en[i]
        target_sentence_ids = val_fr[i]

        # 2. Generate the model's translation
        translated_words = translate_sentence(
            model, source_sentence_ids, en_idx2word, fr_idx2word, device
        )

        # 3. Process the reference (target) sentence
        reference_words = []
        for idx in target_sentence_ids:
            # Skip <SOS> (1), <EOS> (2), and <PAD> (0)
            if idx not in [0, 1, 2]:
                reference_words.append(fr_idx2word[idx])

        outputs.append(translated_words)

        # NLTK expects targets to be a list of lists of references
        targets.append([reference_words])

        # Print the first 3 translations to visually inspect performance
        if i < 3:
            source_words = [en_idx2word[idx] for idx in source_sentence_ids if idx not in [0, 1, 2]]
            print(f"Source: {' '.join(source_words)}")
            print(f"Target: {' '.join(reference_words)}")
            print(f"Model:  {' '.join(translated_words)}\n")

    # Calculate the corpus-level BLEU score using NLTK
    # Note: NLTK expects (references, hypotheses)
    score = corpus_bleu(targets, outputs)

    print(f"BLEU Score on Validation Set: {score * 100:.2f}")
    return score

# Run the evaluation
calculate_bleu(model, val_en, val_fr, en_idx2word, fr_idx2word, device)
```
